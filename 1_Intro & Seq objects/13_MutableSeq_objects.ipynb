```python
from Bio.Seq import Seq
```

```python
my_seq = Seq("GCCATTGTAATGGGCCGCTGAAAGGGTGCCCGA")
```

if you try to edit the sequence

```python
my_seq[5] = "G"
```

to edit ; ought to convert it into a mutable sequence (a MutableSeq object) 

```python
from Bio.Seq import MutableSeq
```

```python
mutable_seq = MutableSeq(my_seq)
mutable_seq
```

to change a sequence object

```python
mutable_seq[5] = "C"
```

```python
mutable_seq
```

to remove a sequence object

```python
mutable_seq
```

```python
mutable_seq.remove("T")
```

(.remove) >>> removes the first T nucleotide

```python
mutable_seq
```

```python
mutable_seq.reverse()
```

```python
mutable_seq
```

Alternatively, you can create a MutableSeq object directly from a string:

mutable_seq = MutableSeq("GCCATTGTAATGGGCCGCTGAAAGGGTGCCCGA")

Once you have finished editing your a MutableSeq object, 
it’s easy to get back to a read-only Seq object should you need to:

```python
mutable_seq
```

```python
new_seq = Seq(mutable_seq)
```

```python
new_seq
```
