per-letter-annotations is similar, letter_annotations is a dictionary like attribute which will let you assign any Python sequence (i.e. a string, list or tuple) which has the same length as the sequence:

```python
simple_seq_r.letter_annotations["phred_quality"] = [40, 40, 38, 30]
```

What the actual numbers mean (Phred scale — higher = more confident):

Score	Meaning
40	99.99% confident this letter is correct
38	~99.98% confident
30	99.9% confident
20	99% confident
10	90% confident

```python
print(simple_seq_r.letter_annotations)
```

So in [40, 40, 38, 30]:

The A and T were read with very high confidence
The G slightly less, but still excellent
The C is the least confident of the four — still good, but if you were scanning for potential errors in this sequence, that last letter is the one worth double-checking first

Where these numbers actually come from: the sequencing machine measures the physical/chemical signal strength when it reads each base. A clean, strong signal → high score. A noisy, ambiguous signal → lower score. So this isn't guessed by a human — it's calculated automatically by the machine itself, base by base, at the moment of sequencing.
